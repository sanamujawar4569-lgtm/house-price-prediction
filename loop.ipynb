```python
l = [
    [1, 2, 3, 4],
    (2, 3, 4, 5),
    (3, 4, 5, 6, 7),
    set([23, 44, 67, 89]),
    {'k1': "sudh", 'k2': 'kumar', 'k3': 'singh', 3: 6, 7: 8},
    ["data", "science"]
]

```

```python
# Q1) Try to extract all the list entities
for i in l:
    if type(i) == list:
        print(i)

```

```python
# Q2) Try to extract all the dictionary entities
for i in l:
    if type(i) == dict:
        print(i)
```

```python
# Q3) Try to extract all the tuple entities
for i in l:
    if type(i) == tuple:
        print(i)

```

```python
# Q4) Try to extract all the numerical data
num = []
for i in l:
    if type(i) == list or type(i) == tuple or type(i) == set:
        for j in i:
            if type(j) == int:
                num.append(j)
    elif type(i) == dict:
        for k, v in i.items():
            if type(k) == int:
                num.append(k)
            if type(v) == int:
                num.append(v)

print(num)

```

```python
# Q5) Try to give the summation of the numerical data
print(sum(num))

```

```python
# Q6) Try to find out multiplication of all the individual collection inside dataset
for i in l:
    mul = 1
    if type(i) in [list, tuple, set]:
        for j in i:
            if type(j) == int:
                mul *= j
    elif type(i) == dict:
        for k, v in i.items():
            if type(k) == int:
                mul *= k
            if type(v) == int:
                mul *= v
    print(i, "→ multiplication =", mul)

```

```python
# Q7) Try to unwrap all the collection inside collection and create a flat list by using a loop in simple code in python
flat = []

for i in l:
    if type(i) in [list, tuple, set]:
        for j in i:
            flat.append(j)
    elif type(i) == dict:
        for k, v in i.items():
            flat.append(k)
            flat.append(v)

print("Flat list:", flat)


```
